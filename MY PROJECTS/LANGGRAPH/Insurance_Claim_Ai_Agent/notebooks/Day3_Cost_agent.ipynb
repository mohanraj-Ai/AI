```python
import pandas as pd
import json
```

```python
repair_db = pd.read_csv("../database/repair_cost.csv")

repair_db.head()
```

```python
vision_output = {
    "damaged_parts": [
        "Hood",
        "Front Bumper",
        "Driver's Side Front Fender",
        "Driver's Side Headlight Assembly",
        "Grille",
        "Radiator Support (likely)",
        "Radiator (likely)"
    ],
    "severity": "Severe",
    "damage_summary": """
The black sedan sustained severe frontal impact damage, primarily concentrated on the driver's side.
The hood is extensively crumpled and deformed.
The front bumper is crushed and fractured on the driver's side.
The driver's side front fender and headlight assembly are completely destroyed and missing,
exposing internal engine bay components.
The front grille is severely damaged and deformed.
The extent of the damage indicates significant structural compromise to the vehicle's front end,
likely affecting the radiator support and other critical components.
"""
}

print(vision_output)
```

```python
parts_cost = 0

for part in vision_output["damaged_parts"]:

    row = repair_db[repair_db["Part"] == part]

    if not row.empty:
        parts_cost += int(row["Cost"].values[0])

print("Parts Cost:", parts_cost)
```

```python
severity = vision_output["severity"]

if severity == "Low":
    labor_cost = 2000

elif severity == "Medium":
    labor_cost = 4000

else:
    labor_cost = 7000

print("Labor Cost:", labor_cost)
```

```python
total_cost = parts_cost + labor_cost

print("Total Cost:", total_cost)
```

```python
cost_output = {
    "parts_cost": parts_cost,
    "labor_cost": labor_cost,
    "total_cost": total_cost
}

print(json.dumps(cost_output, indent=4))
```

```python
def estimate_repair_cost(vision_output, repair_db):

    parts_cost = 0

    for part in vision_output["damaged_parts"]:

        row = repair_db[repair_db["Part"] == part]

        if not row.empty:
            parts_cost += int(row["Cost"].values[0])

    severity = vision_output["severity"]

    if severity == "Low":
        labor_cost = 2000

    elif severity == "Medium":
        labor_cost = 4000

    else:
        labor_cost = 7000

    total_cost = parts_cost + labor_cost

    return {
        "parts_cost": parts_cost,
        "labor_cost": labor_cost,
        "total_cost": total_cost
    }
```

```python
result = estimate_repair_cost(vision_output, repair_db)

print(json.dumps(result, indent=4))
```
