```python
❓ Question 4:
The following Retrieval-Augmented Generation (RAG) model does not retrieve the correct document. Debug the query embedding issue
```

```python
import faiss
import numpy as np
import google.generativeai as genai

genai.configure(api_key="AIzaSyAx4dV6WeTsHXzileVzd0vSVGtwEKQ6LIg")

def get_embedding(text, task_type="retrieval_document"):
    response = genai.embed_content(
        model="models/gemini-embedding-001",  # ← Your available model!
        content=text,
        task_type=task_type
    )
    return np.array(response['embedding'], dtype='float32').reshape(1, -1)

test_embed = get_embedding("test")
d = test_embed.shape[1]  # Auto-detect dimension!
print(f"Embedding dimension: {d}")

index = faiss.IndexFlatL2(d)

documents = [
    "AI is transforming industries",
    "FAISS is great for vector search",
    "Generative models are powerful"
]

print("Indexing documents...")
doc_embeddings = np.vstack([get_embedding(doc) for doc in documents])
index.add(doc_embeddings)
print(f"Total indexed: {index.ntotal}")

query = "How is AI transforming industries?"
query_vector = get_embedding(query, task_type="retrieval_query")

distances, indices = index.search(query_vector, k=1)

retrieved_doc = documents[indices[0][0]]
print(f"\nQuery: {query}")
print(f"Retrieved Document: {retrieved_doc}")
print(f"Distance Score: {distances[0][0]:.4f}")
```
